## 1. Les fonctions de base

Commençons à coder ! Nous avons évoqué les données et les fonctions : commençons par les fonctions. Une fonction est à
comprendre comme une instruction que l'on donne à l'ordinateur. Nous avons déjà utilisé une fonction qui s'appelle
`print()`. L'instruction que l'on donne à l'ordinateur avec cette fonction est d'imprimer (c'est-à-dire d'afficher)
quelque chose à l'écran. En programmation, on a tendance à écrire en anglais : ce sera le cas pour toutes les fonctions
de base de Python.

Décomposons cette fonction dans une cellule de code. Deux choses à savoir avant de se lancer :
- l'ordinateur lit nos instructions **ligne par ligne**, de haut en bas, et les exécute dans cet ordre ;
- quand vous voyez un `#`, cela veut dire que ce qui suit est un **commentaire** : tout ce qui suit un `#` sur une ligne
  est ignoré par l'ordinateur. Cela nous permet de laisser des explications dans notre code.

```python
# Ceci est un commentaire. Dans cette cellule, nous demandons à l'ordinateur d'imprimer
# quelque chose deux fois.

print()
print("Hello world")
```

Décomposons tout ça :
- quand on exécute une fonction, on donne d'abord le **nom** de la fonction (ici `print`) ;
- ce nom est toujours suivi de **parenthèses** (ici, ça donne `print()`) ;
- entre les parenthèses, on peut donner des informations supplémentaires à la fonction : on les appelle des
  **arguments**. Dans le cas de `print`, les arguments sont les choses que l'on souhaite afficher à l'écran.

La première fois que l'on a appelé `print`, on ne lui a pas donné d'argument : l'ordinateur a donc simplement imprimé une
ligne vide. La deuxième fois, on lui a donné l'argument `"Hello world"` : c'est donc ce qu'il a imprimé.

Enfin, on peut donner **plusieurs arguments** à une fonction. Pour cela, on sépare chaque argument par une virgule :

```python
# Pour donner plusieurs arguments, on utilise une virgule
# (qu'on appelle le séparateur d'arguments).
print("Bonjour", "monde")
```

Remarquez que `print` ajoute automatiquement un espace entre ses arguments.

## 2. Les données et les variables

Voyons désormais l'autre élément fondamental de la programmation : les **données**. Nous en avons déjà donné à notre
fonction : `"Hello world"` est une donnée, tout comme `"Bonjour"` et `"monde"`. Ce sont les données que l'on a
données à notre fonction en tant qu'arguments, pour qu'elle travaille avec.

En programmation, quand on parle de données, on parle surtout de **variables**. Une variable est comme une boîte avec une
étiquette : l'étiquette est le **nom** de la variable, et la boîte contient une **donnée**. Quand on **déclare** une
variable, on lui donne donc un nom, puis, après le signe `=`, la donnée qu'elle doit contenir :

```python
# Ici, je déclare une variable que j'appelle ma_variable.
# Je dis qu'elle est égale à la donnée "Hello world, je suis contenu dans une variable." :
ma_variable = "Hello world, je suis contenu dans une variable."

# Maintenant, au lieu de donner la donnée directement comme argument à une fonction,
# je peux donner le nom de la variable :
print(ma_variable)
```

Remarquez que c'est la donnée contenue dans la variable qui est imprimée, et non son nom.

> 💡 Quelques règles pour les noms de variables : pas d'espaces (on utilise `_` à la place), pas d'accents, et le nom ne
> doit pas commencer par un chiffre. Choisissez des noms qui décrivent le contenu : `nom_auteur` est bien plus clair
> que `x` !

Pourquoi utiliser des variables plutôt que de donner la donnée directement ? Il y a plusieurs bonnes raisons, que nous
comprendrons au fur et à mesure. Prenons un premier exemple.

Ici, je donne une nouvelle valeur à ma variable (de la même manière que je l'ai déclarée), puis je l'imprime plusieurs
fois. Je pourrais copier-coller mon texte directement dans chaque `print`, mais que se passe-t-il si je veux modifier ce
texte ? Je serais obligé de changer l'argument de chaque `print`, ce qui est pénible et source d'erreurs. Avec une
variable, il suffit de la modifier une seule fois, et chaque `print` utilisera automatiquement la nouvelle valeur.

```python
ma_variable = "Mon texte super compliqué que je ne dois écrire qu'une fois au lieu de six !"

print(ma_variable)
print(ma_variable)
print(ma_variable)
print(ma_variable)
print(ma_variable)
print(ma_variable)
```

## 3. Les types de données

En programmation, il existe plusieurs types de données. Pour l'instant, nous n'en avons vu qu'un : les **strings**
(chaînes de caractères), qui sont des suites de caractères écrites entre guillemets. Bien entendu, il existe d'autres
types de données, plus ou moins complexes.

### Les données simples : string, integer, float et boolean

- Les **strings** (`str`) : des suites de caractères, du texte. Ils s'écrivent entre guillemets doubles (`""`) ou
  simples (`''`) : c'est la même chose.
- Les **integers** (`int`) : des nombres entiers. On écrit simplement le nombre.
- Les **floats** (`float`) : des nombres décimaux. Attention, Python utilise la notation anglo-saxonne : on écrit un
  **point** `.` à la place de la virgule.
- Les **booleans** (`bool`) : une valeur qui n'a que deux états possibles, `True` (vrai) ou `False` (faux). Attention à
  la majuscule, et pas de guillemets !

Créons quelques variables de chacun de ces types, et imprimons-les :

```python
# Un string, une suite de caractères entre "" :
mon_string_1 = "Hello world"

# Ou bien entre '' :
mon_string_2 = 'Hello world'

# Un integer : on écrit simplement un nombre entier :
mon_int = 20

# Un float : un nombre décimal (avec un point et non une virgule) :
mon_float = 20.5

# Un boolean : soit True, soit False :
mon_bool = False

print(mon_string_1)
print(mon_string_2)
print(mon_int)
print(mon_float)
print(mon_bool)
```

### La fonction `type()`

Il existe une fonction de base qui nous permet de connaître le type de la donnée contenue dans une variable :
`type()`. On lui donne en argument la variable dont on souhaite connaître le type.

Notons que :
- les variables déclarées dans la cellule précédente existent toujours dans les cellules suivantes du notebook (à
  condition d'avoir exécuté la cellule précédente !) ;
- on peut mettre directement le résultat d'une fonction comme argument d'une autre : ici, on met `type()` à l'intérieur
  de `print()`.

```python
# On peut récupérer le résultat de la fonction type et le mettre dans une nouvelle variable :
le_type_de_mon_string = type(mon_string_1)
print(le_type_de_mon_string)

# Ou bien mettre la fonction type directement comme argument de la fonction print :
print(type(mon_string_2))
print(type(mon_int))
print(type(mon_float))
print(type(mon_bool))
```

### Opérations sur les données

On peut faire des calculs avec les nombres (integers et floats), avec les opérateurs habituels : `+` (addition), `-`
(soustraction), `*` (multiplication) et `/` (division). Comme en maths, on peut utiliser des parenthèses pour indiquer
les priorités.

Certains opérateurs fonctionnent aussi avec les strings : `+` colle deux strings l'un à la suite de l'autre (on appelle
ça une **concaténation**).

Attention, on ne peut pas mélanger n'importe quoi : `"20" + 1` provoque une erreur, car `"20"` est un string (du texte),
pas un nombre ! On peut convertir une donnée d'un type à un autre avec les fonctions `int()`, `float()` et `str()`.

```python
print(10 + 5)
print(10 - 5)
print(10 * 5)
print(10 / 4)          # une division donne toujours un float
print((2 + 3) * 4)     # les parenthèses sont calculées en premier

prenom = "Victor"
nom = "Hugo"
print(prenom + " " + nom)   # concaténation de strings

print(int("20") + 1)        # on convertit le string "20" en integer avant de faire le calcul
print("J'ai " + str(20) + " ans")  # et ici l'integer 20 en string
```

### Les données plus complexes : listes et dictionnaires

Il existe d'autres types de données, qui permettent de regrouper plusieurs données en une seule. Nous allons en voir
deux, très utilisés.

#### Les listes

Une **liste** (`list`) permet de regrouper plusieurs données, quel que soit leur type (par exemple une chaîne de
caractères avec des nombres). Une liste s'écrit entre crochets `[]`, et on sépare chaque élément par une virgule :

```python
# Création d'une liste : on écrit entre crochets, et on sépare chaque élément par une virgule.
ma_liste = [1, 2, 3.4, 4, True, "hello"]
```

On peut afficher une liste de deux manières :
1. **en entier** : il suffit de donner le nom de la liste à `print` (instruction n°1 dans la cellule suivante) ;
2. **un seul élément** : on donne le nom de la liste, suivi entre crochets `[]` de la **position** de l'élément qui nous
   intéresse. Par exemple, `ma_liste[2]` pour obtenir le float `3.4` (instruction n°2).

⚠ **Remarque importante** : en programmation, on **compte à partir de 0**, et non à partir de 1. Pour récupérer le
premier élément d'une liste, on donne donc la position 0 ; pour le deuxième, la position 1, et ainsi de suite. La
position d'un élément dans une liste s'appelle son **index**.

```python
# On peut imprimer la liste en entier :
print(ma_liste) # instruction n°1

# Ou bien un élément précis, d'après sa position (son index).
# N'oubliez pas que l'on compte à partir de 0 :
print(ma_liste[2]) # instruction n°2
```

#### Les dictionnaires

Un **dictionnaire** (`dict`) ressemble à une liste, mais au lieu d'être repérée par sa position, chaque valeur est
repérée par une étiquette qui lui est propre, appelée **clé** (le plus souvent un string). Un dictionnaire est donc un
ensemble de paires *clé : valeur*, comme un vrai dictionnaire associe une définition à chaque mot.

Un dictionnaire s'écrit entre accolades `{}`, et, comme pour les listes, on sépare chaque paire par une virgule. Chaque
paire s'écrit de la même manière : la clé, le symbole `:`, puis la valeur (instruction n°1 dans la cellule suivante).

Comme pour les listes, on peut afficher un dictionnaire en entier (instruction n°2), ou récupérer une valeur précise.
Mais au lieu de donner un index, on donne la clé qui nous intéresse : par exemple `"element1"` pour obtenir la valeur
`5` (instruction n°3).

Remarque : on récupère les éléments d'un dictionnaire avec des crochets `[]`, comme pour les listes, et non avec des
accolades `{}`, même si cela pourrait sembler plus logique.

```python
# Je déclare un dictionnaire entre accolades, avec une suite de paires clé : valeur.
# On peut revenir à la ligne entre chaque élément pour rendre le tout plus lisible :
mon_dict = {
  "element1": 5,
  "element2": "hello world"
} # instruction n°1

# On peut imprimer le dictionnaire en entier :
print(mon_dict) # instruction n°2

# Ou bien trouver une valeur précise avec sa clé :
print(mon_dict["element1"]) # instruction n°3
```

### Mettre à jour les valeurs des variables

Dernière chose sur les variables : on peut changer la donnée contenue dans une variable à tout moment, de la même
manière qu'on l'a déclarée.

Pour les listes et les dictionnaires, on peut aussi modifier un élément précis, avec la même syntaxe que pour y
accéder. Et pour un dictionnaire, si la clé n'existe pas encore, elle est ajoutée !

```python
# Changer la valeur d'une variable simple.
# Ma variable a déjà une valeur :
print(mon_int)
# Ici, je change sa valeur :
mon_int = 10
print(mon_int)
print("--")

# Les listes.
print(ma_liste)
print(ma_liste[0])
# Ici, je change le premier élément de ma liste :
ma_liste[0] = 6
print(ma_liste)
print(ma_liste[0])
print("--")

# Même opération pour les dictionnaires :
print(mon_dict)
print(mon_dict["element1"])
mon_dict["element1"] = "Une autre valeur"
print(mon_dict)
print(mon_dict["element1"])

# Et si la clé n'existe pas encore, elle est créée :
mon_dict["element3"] = 42
print(mon_dict)
```

## 4. Créer ses propres fonctions

Ça y est, nous avons vu les fondamentaux : les données et les fonctions ! Mais si l'on s'en tient aux fonctions de base
de Python, on va rapidement être limité. On va vite vouloir créer nos propres fonctions, pour faire faire à
l'ordinateur des tâches plus complexes et plus adaptées à nos besoins.

Pour exécuter une fonction, on donnait son nom, des parenthèses, et éventuellement des arguments entre ces
parenthèses. Pour **créer** une fonction, on reprend la même syntaxe, mais précédée du mot clé `def` (pour *define*).

> 💡 Un **mot clé** (ou mot réservé) est un mot que Python utilise pour comprendre ce que l'on veut faire. On ne peut
> donc pas l'utiliser comme nom de variable ou de fonction.

On écrit donc le mot clé `def`, le nom que l'on veut donner à notre fonction, puis des parenthèses avec des noms
d'arguments. Comme on est en train de *créer* la fonction, on ne donne pas de données entre les parenthèses : on
invente des noms de variables, qui n'existeront qu'à l'intérieur de la fonction, et qui seront remplacés par les
arguments donnés au moment où l'on exécutera la fonction. Enfin, on met deux points `:`, et on écrit le contenu de la
fonction à la ligne, en décalant chaque ligne vers la droite.

Voyons cela en détail :

```python
# Ici, je crée une fonction qui s'appelle imprimer_a_lecran.
# J'écris def, le nom de la fonction, des parenthèses avec des noms d'arguments,
# puis deux points :

def imprimer_a_lecran(argument_1, un_autre_argument):
  # Ce qui suit est le contenu de ma fonction :
  print(argument_1)
  print(un_autre_argument)
```

Quand on exécute cette cellule, il ne se passe rien ! C'est normal : nous avons seulement *déclaré* notre fonction. On a
dit à Python qu'elle existe, mais on ne l'a pas *exécutée*. Pour cela, on donne son nom et les arguments que l'on
souhaite lui passer :

```python
imprimer_a_lecran("Bonjour tout le monde", "Comment ça va ?")
imprimer_a_lecran("Ça va très bien", "Et toi ?")
```

### L'indentation

Nous avons créé notre première fonction ! Remarquez que tout le contenu de notre fonction `imprimer_a_lecran()` était
écrit à la ligne, mais aussi **indenté**, c'est-à-dire décalé vers la droite (avec la touche **Tab**, ou deux ou quatre
espaces).

L'indentation est très importante en Python. En général, dès qu'il y a deux points `:`, comme dans la déclaration de
notre fonction, cela veut dire que ce qui suit doit être indenté, et que tout ce qui est indenté appartient à l'élément
qui se termine par les deux points. Dans le cas d'une fonction : on la déclare, on met deux points, puis tout ce qui est
indenté appartient à la fonction.

Cela nous permet par exemple de déclarer une fonction, d'écrire son contenu, puis de continuer avec du code qui n'est
plus indenté, et qui ne fait donc plus partie de la fonction. Voici un exemple :

```python
# Je crée d'abord une variable au niveau d'indentation 0 :
my_var = 20

# Puis je déclare une fonction :
def ajouter_10(valeur):
  # Toutes les lignes indentées appartiennent à la fonction.

  # D'abord, j'imprime la variable my_var, déclarée au niveau d'indentation 0 :
  print(my_var)

  # J'ajoute 10 à la valeur, et je mets le résultat dans une variable qui s'appelle resultat :
  resultat = valeur + 10

  # Puis j'imprime le résultat :
  print(resultat)

# Maintenant, je reviens au niveau d'indentation 0 pour exécuter ma fonction :
ajouter_10(my_var)
```

Une variable déclarée à un niveau d'indentation est accessible aux niveaux plus indentés, mais pas aux niveaux
précédents. Qu'est-ce que ça veut dire ? Dans cet exemple, `my_var` est déclarée au niveau 0 : elle est donc accessible
à l'intérieur de la fonction. En revanche, la variable `resultat` est déclarée au niveau 1, à l'intérieur de la
fonction : elle n'existe que dans la fonction. Si j'essayais d'imprimer `resultat` au niveau 0, après la dernière ligne,
j'obtiendrais une erreur m'indiquant que la variable n'existe pas (on appelle cela la **portée** d'une variable).

### Renvoyer un résultat avec `return`

Alors, comment récupérer le résultat calculé par une fonction ? Avec le mot clé `return`. Quand Python rencontre
`return`, il sort de la fonction et **renvoie** la valeur indiquée. On peut alors mettre cette valeur dans une variable,
et continuer à travailler avec.

Attention à la différence entre `print` et `return` : `print` *affiche* une valeur à l'écran, mais elle est ensuite
perdue ; `return` *renvoie* la valeur au code qui a appelé la fonction.

```python
def ajouter_10_et_renvoyer(valeur):
  resultat = valeur + 10
  return resultat

# La valeur renvoyée par la fonction est mise dans la variable mon_resultat :
mon_resultat = ajouter_10_et_renvoyer(5)
print(mon_resultat)

# On peut donc réutiliser ce résultat, par exemple dans un autre appel de la fonction :
print(ajouter_10_et_renvoyer(mon_resultat))
```
