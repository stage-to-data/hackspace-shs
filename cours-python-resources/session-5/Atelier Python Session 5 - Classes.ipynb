Aujourd'hui, nous allons découvrir la **programmation orientée objet** (POO, ou OOP en anglais pour *Object-Oriented
Programming*). C'est une manière d'organiser son code en regroupant, dans un même « objet », des **données** et les
**fonctions** qui travaillent sur ces données : les deux notions fondamentales de la séance 1, réunies !

En fait, vous utilisez déjà des objets depuis plusieurs séances sans le savoir :
- un fichier ouvert avec `open()` est un objet, avec des méthodes comme `read()` ou `close()` ;
- la réponse de `requests.get()` est un objet, avec des attributs comme `status_code` ou `text` ;
- la « soupe » de `BeautifulSoup` est un objet, avec des méthodes comme `find_all()` ;
- et même les listes (`ma_liste.append()`) et les strings (`mon_texte.lower()`) sont des objets !

Aujourd'hui, nous allons apprendre à créer nos propres types d'objets.

## 1. Classes, instances, attributs et méthodes

Quelques mots de vocabulaire :
- une **classe** est un modèle, un « moule », qui décrit un type d'objet. Par exemple : « un étudiant a un prénom, un
  nom et des notes, et on peut calculer sa moyenne » ;
- une **instance** est un objet concret fabriqué à partir de ce moule : l'étudiante Clarisse, l'étudiant Jacob... On
  peut créer autant d'instances que l'on veut à partir d'une même classe ;
- les **attributs** sont les données propres à chaque instance (le prénom, les notes...) ;
- les **méthodes** sont les fonctions définies dans la classe, qui travaillent sur les attributs de l'instance (calculer
  la moyenne...).

Pourquoi est-ce utile ? Imaginons que l'on veuille gérer les notes d'une classe entière. Sans objets, on pourrait
utiliser un dictionnaire par étudiant, et écrire des fonctions à côté. Avec une classe, les données et les fonctions qui
vont ensemble sont rangées au même endroit, ce qui rend le code plus clair et plus facile à réutiliser.

## 2. Créer une classe

Pour créer une classe, on utilise le mot clé `class`, suivi du nom de la classe (par convention, avec une majuscule, et
sans tirets bas : `Etudiant`, `EtudiantSpeAnglais`), puis deux points.

À l'intérieur, on définit une méthode spéciale, `__init__()` (avec deux tirets bas de chaque côté). Elle est appelée
automatiquement à chaque fois que l'on crée une nouvelle instance : c'est elle qui **initialise** l'objet, en donnant
leurs valeurs de départ aux attributs.

Son premier argument est toujours `self` : il représente l'instance elle-même, celle qui est en train d'être créée.
`self.prenom = prenom_etudiant` veut dire : « range la valeur de l'argument `prenom_etudiant` dans l'attribut `prenom`
de cette instance ».

```python
# Pour créer une classe, on utilise le mot clé class, puis son nom (avec une majuscule) :
class Etudiant:

  # La méthode __init__ prend toujours au moins l'argument self :
  def __init__(self, prenom_etudiant: str, nom_de_famille: str, math: float, francais: float):
    """Représente un étudiant, avec ses notes de maths et de français."""
    self.prenom = prenom_etudiant
    self.nom_de_famille = nom_de_famille
    self.note_maths = math
    self.note_francais = francais
```

Deux petites nouveautés dans ce code, qui ne sont pas obligatoires mais sont de bonnes pratiques :
- les **annotations de type** : `prenom_etudiant: str` indique que l'on attend un string pour cet argument, et
  `math: float` un nombre. Python ne vérifie rien, mais cela aide à comprendre le code (et Colab s'en sert pour ses
  suggestions) ;
- la **docstring** : le texte entre triples guillemets `"""..."""` juste sous la ligne `def` décrit ce que fait la
  fonction. Elle s'affiche quand on survole la fonction dans Colab, ou avec `help()`.

## 3. Créer des instances

Pour créer une instance, on appelle la classe comme une fonction, en lui donnant les arguments de `__init__` (sauf
`self`, que Python fournit tout seul) :

```python
etudiant_1 = Etudiant("Jacob", "Hart", 12, 10)
etudiant_2 = Etudiant("Clarisse", "Bardiot", 16, 20)

print(etudiant_1.note_francais)
print(etudiant_2.note_francais)
```

`etudiant_1` et `etudiant_2` sont deux instances de la même classe : elles ont les mêmes attributs, mais avec des
valeurs différentes. On accède à un attribut avec le nom de l'instance, un point, puis le nom de l'attribut. On peut
aussi modifier un attribut, comme une variable :

```python
print(etudiant_1.prenom)

etudiant_1.note_maths = 14
print(etudiant_1.note_maths)

print(type(etudiant_1))  # le type d'une instance, c'est sa classe !
```

## 4. Les méthodes

Imaginons que l'on veuille calculer la moyenne de chaque étudiant. On pourrait le faire à l'extérieur de la classe :

```python
moyenne_1 = (etudiant_1.note_francais + etudiant_1.note_maths) / 2
moyenne_2 = (etudiant_2.note_francais + etudiant_2.note_maths) / 2

print(moyenne_1)
print(moyenne_2)
```

Mais il faut réécrire le calcul pour chaque étudiant, et si la formule change, il faudra la modifier partout. Il est
bien plus pratique de l'écrire une seule fois, **dans la classe**, sous forme de méthode.

Une méthode se définit comme une fonction, avec `def`, mais à l'intérieur de la classe (donc indentée), et avec `self`
comme premier argument. Grâce à `self`, la méthode peut accéder aux attributs de l'instance sur laquelle on l'appelle.

Ajoutons deux méthodes à notre classe :
- `get_moyenne()`, qui **renvoie** la moyenne de l'étudiant ;
- `__str__()`, une autre méthode spéciale, qui est appelée quand on fait `print()` d'une instance : elle doit renvoyer
  un string qui décrit l'objet. Sans elle, `print(etudiant_1)` affiche quelque chose de peu parlant comme
  `<__main__.Etudiant object at 0x7f...>`.

Comme on modifie la classe, il faut réexécuter sa définition, puis **recréer** les instances.

```python
class Etudiant:

  def __init__(self, prenom_etudiant: str, nom_de_famille: str, math: float, francais: float):
    """Représente un étudiant, avec ses notes de maths et de français."""
    self.prenom = prenom_etudiant
    self.nom_de_famille = nom_de_famille
    self.note_maths = math
    self.note_francais = francais

  def get_moyenne(self):
    """Renvoie la moyenne des notes de l'étudiant."""
    moyenne = (self.note_maths + self.note_francais) / 2
    return moyenne

  def __str__(self):
    return f"{self.prenom} {self.nom_de_famille} (moyenne : {self.get_moyenne()})"


etudiant_1 = Etudiant("Jacob", "Hart", 12, 10)
etudiant_2 = Etudiant("Clarisse", "Bardiot", 16, 20)
```

On appelle une méthode comme on accède à un attribut, mais avec des parenthèses, comme pour une fonction (et, là
encore, sans donner `self`) :

```python
moyenne_1 = etudiant_1.get_moyenne()
moyenne_2 = etudiant_2.get_moyenne()

print(moyenne_1)
print(moyenne_2)

print(etudiant_1)  # appelle automatiquement __str__
```

Remarquez que dans `__str__`, on appelle une autre méthode de la classe avec `self.get_moyenne()`.

Les méthodes peuvent aussi **modifier** les attributs de l'instance. On peut même stocker d'autres objets dans les
attributs, par exemple une liste d'étudiants dans une classe `Promotion` :

```python
class Promotion:

  def __init__(self, nom: str):
    self.nom = nom
    self.etudiants = []  # une liste vide, propre à chaque promotion

  def ajouter(self, etudiant: Etudiant):
    """Ajoute un étudiant à la promotion."""
    self.etudiants.append(etudiant)

  def get_moyenne_generale(self):
    """Renvoie la moyenne des moyennes des étudiants."""
    total = 0
    for etudiant in self.etudiants:
      total = total + etudiant.get_moyenne()
    return total / len(self.etudiants)


promo = Promotion("Master Humanités numériques")
promo.ajouter(etudiant_1)
promo.ajouter(etudiant_2)

print(len(promo.etudiants), "étudiants dans la promotion", promo.nom)
print("Moyenne générale :", promo.get_moyenne_generale())
```

## 5. L'héritage

Imaginons maintenant que certains étudiants suivent une spécialité anglais : ils ont une note de plus, et leur moyenne
se calcule différemment. Plutôt que de recopier toute la classe `Etudiant`, on peut créer une nouvelle classe qui
**hérite** d'`Etudiant` : elle récupère automatiquement tous ses attributs et toutes ses méthodes, et on n'a plus qu'à
écrire ce qui change.

Pour cela, on donne la classe « parente » entre parenthèses après le nom de la nouvelle classe :
`class EtudiantSpeAnglais(Etudiant):`.

Dans la classe « enfant », la fonction `super()` permet d'accéder aux méthodes de la classe parente :
- dans `__init__`, `super().__init__(...)` exécute le `__init__` d'`Etudiant`, qui s'occupe du prénom, du nom et des
  deux premières notes ; il ne reste plus qu'à ajouter la note d'anglais ;
- en redéfinissant `get_moyenne()`, on **remplace** la méthode de la classe parente. Mais on peut toujours appeler
  l'ancienne version avec `super().get_moyenne()`, pour ne pas réécrire le calcul.

```python
class EtudiantSpeAnglais(Etudiant):

  def __init__(self, prenom_etudiant: str, nom_de_famille: str, math: float, francais: float, anglais: float):
    # On appelle le __init__ de la classe parente (Etudiant) :
    super().__init__(prenom_etudiant, nom_de_famille, math, francais)
    # Puis on ajoute ce qui est propre à cette classe :
    self.note_anglais = anglais

  def get_moyenne(self):
    """Renvoie la moyenne des trois notes."""
    # super().get_moyenne() est la moyenne des maths et du français, calculée par la classe parente.
    # On la multiplie par 2 pour retrouver la somme des deux notes, et on ajoute la note d'anglais :
    moyenne_parente = super().get_moyenne()
    return ((moyenne_parente * 2) + self.note_anglais) / 3


etudiant_3 = EtudiantSpeAnglais("Jeanne", "Fras", 9, 14, 18)

print(etudiant_3.nom_de_famille)  # attribut défini par le __init__ d'Etudiant
print(etudiant_3.note_anglais)    # attribut défini par le __init__ d'EtudiantSpeAnglais
print(etudiant_3.get_moyenne())   # la nouvelle version de get_moyenne
print(etudiant_3)                 # __str__ est hérité d'Etudiant, et utilise la nouvelle get_moyenne !
```

Un `EtudiantSpeAnglais` est aussi un `Etudiant` : on peut donc l'ajouter à une `Promotion`, et la moyenne générale
utilisera automatiquement la bonne formule pour chaque étudiant.

```python
promo.ajouter(etudiant_3)
print("Moyenne générale :", promo.get_moyenne_generale())
```
