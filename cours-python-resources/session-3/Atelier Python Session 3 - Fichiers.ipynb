## 1. Le package `os`

Nous avons terminé la dernière séance en important le package `random` pour produire des nombres aléatoires.

```python
import os
```

Le package `os` nous permet de gérer tout ce qui concerne les dossiers et les fichiers de notre ordinateur. Voyons

```python
print(os.getcwd())
```

Quand on manipule des fichiers, il faut imaginer Python qui se déplace dans l'arborescence des dossiers de

- un **chemin relatif** : la position du fichier par rapport au **dossier de travail** de Python (par exemple

`os.getcwd()` nous donne justement ce dossier de travail. Dans Google Colab, c'est un dossier qui s'appelle `/content` :

### Le chemin vers un fichier

Pour dire à Python où trouver un fichier, il faut lui donner son chemin. On peut par exemple écrire le chemin vers le
fichier `README.md` du dossier `sample_data` de la manière suivante :

```python
chemin = "/content/sample_data/README.md"
```

Cela fonctionne ici, mais ce n'est pas une bonne pratique. En effet, sous Windows, les chemins ne s'écrivent pas de la
même manière (on utilise `\` à la place de `/`). Il vaut donc mieux utiliser le package `os` pour construire nos
chemins : il s'occupe des particularités de chaque système d'exploitation, et notre script fonctionnera partout.

Pour cela, on utilise la fonction `os.path.join()` (`path` est un sous-module de `os`), qui assemble les morceaux
d'un chemin :

```python
# Mettons la racine dans une variable, pour n'exécuter os.getcwd() qu'une seule fois :
racine = os.getcwd()

# Construisons le chemin en donnant chaque morceau comme argument à os.path.join :
chemin = os.path.join(racine, "sample_data", "README.md")
print(chemin)
```

On peut vérifier que Python trouve bien le fichier avec la fonction `os.path.isfile()`, qui renvoie un boolean :
`True` si le fichier existe, `False` sinon. On peut faire la même chose avec les dossiers grâce à `os.path.isdir()`.

```python
print(os.path.isfile(chemin))
print(os.path.isdir(os.path.join(racine, "sample_data")))
print(os.path.isfile(os.path.join(racine, "sample_data", "fichier_qui_nexiste_pas.txt")))
```

## 2. Lire des fichiers

### Les fichiers texte

Super ! Nous savons indiquer à Python où trouver un fichier. Maintenant, lisons-le ! Pour cela, on utilise une fonction
qui fait partie de Python, `open()`.

On met le résultat de `open()` dans une variable. Ce résultat est un objet « fichier ouvert » : on peut l'utiliser avec
des **méthodes** (on écrit le nom de la variable, un point, puis la méthode, comme `ma_liste.append()` en séance 2).
Nous reviendrons plus en détail sur ces objets lors de la séance sur les classes.

Voici comment ouvrir et lire le contenu d'un fichier texte, comme le fichier `.md` défini plus haut :

```python
# D'abord, on ouvre le fichier avec open(), en donnant le chemin du fichier en argument.
# On précise aussi l'encodage du fichier (voir plus bas) :
f = open(chemin, encoding="utf-8")

# Puis on lit le contenu du fichier avec la méthode read(), et on le met dans une variable :
data = f.read()

# Enfin, on ferme le fichier, qui est toujours ouvert, avec la méthode close() :
f.close()

print(data)
```

Notez que l'on **ouvre** d'abord le fichier, puis qu'on **lit** son contenu : ce n'est pas la même chose ! Et il ne
faut pas oublier de **fermer** le fichier, pour qu'il n'occupe plus de place dans la mémoire vive de l'ordinateur.

> 💡 **L'encodage** : un fichier texte est enregistré sous forme de nombres, et l'encodage est la « table de
> correspondance » entre ces nombres et les caractères. L'encodage le plus courant aujourd'hui est `utf-8`. Précisez-le
> toujours avec `encoding="utf-8"` : sans cela, sous Windows, les accents risquent d'être mal lus (`dÃ¨s` au lieu de
> `dès`).

Voilà ! Nous avons ouvert et lu un fichier en très peu de lignes de code. Le contenu est un simple string : on peut donc
utiliser tout ce que l'on sait sur les strings, par exemple compter ses caractères avec `len()`, ou le parcourir avec
une boucle `for` (qui donne un caractère à chaque tour).

Les strings ont aussi des méthodes très utiles pour travailler avec du texte :

```python
print(len(data), "caractères")

premiere_phrase = "This directory includes a few sample datasets to get you started."

print(premiere_phrase.lower())          # tout en minuscules
print(premiere_phrase.upper())          # tout en majuscules
print(premiere_phrase.split())          # découpe en une liste de mots (sur les espaces)
print(data.split("\n"))                 # découpe sur les sauts de ligne ("\n")
print(premiere_phrase.count("a"))       # compte les occurrences de "a"
print("started.".strip("."))            # enlève les caractères indiqués au début et à la fin
print(premiere_phrase.replace("few", "lot of"))  # remplace un morceau de texte par un autre
```

### Les fichiers structurés : JSON

Cette méthode est très bien pour lire des fichiers texte, dont on veut traiter le contenu comme un string, c'est-à-dire
une suite de caractères. Mais on a souvent besoin de lire des fichiers de **données structurées**, comme des fichiers
**JSON** ou **CSV**.

Voyons d'abord les fichiers JSON. Le dossier `sample_data` contient un fichier qui s'appelle `anscombe.json`.
Ouvrez-le dans Colab (double-cliquez dessus dans le panneau des fichiers) pour voir son contenu, et définissons son
chemin comme tout à l'heure :

```python
chemin = os.path.join(racine, "sample_data", "anscombe.json")
```

Si vous regardez le contenu de ce fichier, vous verrez qu'il ressemble beaucoup à quelque chose que nous connaissons :
des listes et des dictionnaires Python ! En effet, le format [JSON](https://fr.wikipedia.org/wiki/JavaScript_Object_Notation)
(*JavaScript Object Notation*) utilise presque la même syntaxe : des objets entre accolades `{}` avec des paires
clé : valeur (nos dictionnaires), et des tableaux entre crochets `[]` (nos listes). C'est l'un des formats les plus
utilisés pour échanger des données, notamment sur le web.

On peut lire un fichier JSON comme un fichier texte, mais on obtient alors un simple string. Cela ne nous permet pas de
faire des opérations comme retrouver un élément précis (par exemple avec `mon_dict["nom_de_cle"]`).

Pour cela, il faut **décoder** ce contenu textuel pour le transformer en listes et dictionnaires Python. On utilise un
autre package, qui s'appelle `json` (il fait partie de Python), et sa fonction `json.loads()` (*load string*) fait
exactement cela.

*(Remarque : le fichier `anscombe.json` contient une **liste** de dictionnaires, et non un seul dictionnaire. C'est tout
à fait valide en JSON : un fichier JSON peut contenir un objet `{}` ou une liste `[]`.)*

```python
# D'abord, on importe le package json :
import json

# Puis on lit notre fichier comme avant :
f = open(chemin, encoding="utf-8")
data = f.read()
f.close()

print(type(data))  # c'est un string

# Cette fois-ci, on utilise json.loads() pour transformer le string en données Python :
data_json = json.loads(data)
print(type(data_json))  # c'est une liste

# Maintenant, on peut travailler avec comme avec de vraies données structurées.
# Chaque élément de la liste est un dictionnaire, qui a notamment une clé "X" :
for point in data_json:
  print(point["X"])
```

Voilà comment lire les fichiers JSON. Il existe aussi une manière plus rapide, qui évite de passer par le texte : la
fonction `json.load()` (sans `s`), qui prend directement le fichier ouvert en argument.

```python
# On peut aussi lire directement les données structurées :
f = open(chemin, encoding="utf-8")
data = json.load(f)
f.close()

print(data[0])
print(len(data), "points")
```

### Les fichiers structurés : CSV

L'autre grand format de données structurées est le **CSV** (*Comma-Separated Values*, valeurs séparées par des
virgules). C'est un **tableau** enregistré sous forme de texte : chaque ligne du fichier est une ligne du tableau, et
les cellules sont séparées par des virgules (parfois par des points-virgules ou des tabulations). La première ligne
contient souvent les noms des colonnes : c'est l'**en-tête**. On peut ouvrir un CSV dans un tableur comme Excel ou
LibreOffice Calc.

Le dossier `sample_data` contient le fichier `california_housing_train.csv`, qui décrit des quartiers de Californie
(données du recensement de 1990 : position, âge des logements, population, revenu médian, prix médian des maisons...).

Pour le lire, on utilise le package `csv` (qui fait partie de Python) et sa fonction `csv.reader()`. Elle prend en
argument le fichier ouvert, et renvoie un objet que l'on peut parcourir avec une boucle `for` : à chaque tour, on
obtient une ligne du tableau, sous forme de **liste** de cellules.

```python
import csv

chemin = os.path.join(racine, "sample_data", "california_housing_train.csv")

lignes = []

f = open(chemin, encoding="utf-8")
lecteur = csv.reader(f, delimiter=",")  # delimiter indique le séparateur des cellules
for ligne in lecteur:
  lignes.append(ligne)
f.close()

print(len(lignes), "lignes")

# La première ligne est l'en-tête, avec le nom des colonnes :
print(lignes[0])

# Les lignes suivantes sont les données :
print(lignes[1])
```

On obtient donc une **liste de listes** : `lignes[1]` est la deuxième ligne du tableau, et `lignes[1][3]` la quatrième
cellule de cette ligne.

⚠ **Attention** : `csv.reader()` lit toutes les cellules comme des **strings**, même quand elles contiennent des
nombres ! Pour faire des calculs, il faut donc les convertir avec `float()` ou `int()`.

Calculons par exemple le prix médian moyen des maisons (la dernière colonne, `median_house_value`). Pour ne pas
traiter l'en-tête comme une ligne de données, on utilise `enumerate()` et on saute la ligne d'index 0 :

```python
# L'en-tête nous dit à quel index se trouve la colonne qui nous intéresse :
index_prix = lignes[0].index("median_house_value")  # la méthode index() donne la position d'un élément dans une liste
print("La colonne median_house_value est à l'index", index_prix)

total = 0
nb_quartiers = 0
for numero, ligne in enumerate(lignes):
  if numero > 0:  # on saute l'en-tête
    total = total + float(ligne[index_prix])
    nb_quartiers = nb_quartiers + 1

print("Prix médian moyen :", round(total / nb_quartiers), "dollars")
```

> 💡 Il existe aussi `csv.DictReader()`, qui utilise l'en-tête pour renvoyer chaque ligne sous forme de dictionnaire :
> `ligne["median_house_value"]` au lieu de `ligne[8]`. Et lors de la séance sur la visualisation de données, nous
> verrons le package `pandas`, qui rend la manipulation de tableaux encore plus simple.

## 3. Écrire des fichiers

Lire des fichiers, c'est bien ; en créer, c'est aussi très utile, par exemple pour enregistrer les résultats de nos
analyses !

### Écrire un fichier texte

Pour écrire dans un fichier, on utilise aussi `open()`, mais avec un deuxième argument : le **mode** d'ouverture.
- `"r"` (*read*, lecture) : c'est le mode par défaut, celui que nous avons utilisé jusqu'ici ;
- `"w"` (*write*, écriture) : crée le fichier s'il n'existe pas. ⚠ **S'il existe déjà, son contenu est effacé !** ;
- `"a"` (*append*, ajout) : ajoute du texte à la fin du fichier, sans effacer ce qu'il contient déjà.

Puis on utilise la méthode `write()`, qui prend en argument le string à écrire :

```python
mon_texte = "Bonjour, comment ça va ?\nTrès bien, merci !\n"

chemin = os.path.join(racine, "sample_data", "nouveau_fichier.txt")

f = open(chemin, "w", encoding="utf-8")
f.write(mon_texte)
f.close()

# On ajoute une ligne à la fin du fichier avec le mode "a" :
f = open(chemin, "a", encoding="utf-8")
f.write("Une ligne ajoutée à la fin.\n")
f.close()

# Vérifions en relisant le fichier :
f = open(chemin, encoding="utf-8")
print(f.read())
f.close()
```

Remarquez le `\n` dans le texte : c'est un caractère spécial qui représente un **saut de ligne**. `write()` n'ajoute
pas de saut de ligne tout seul, contrairement à `print()`.

Ouvrez le panneau des fichiers de Colab (vous devrez peut-être cliquer sur 🔄 pour rafraîchir) : le fichier
`nouveau_fichier.txt` est apparu dans `sample_data` !

Comme le mode `"w"` efface les fichiers existants, on peut se protéger avec une condition :

```python
if os.path.isfile(chemin):
  print("Attention, ce fichier existe déjà ! Je ne l'écrase pas.")
else:
  f = open(chemin, "w", encoding="utf-8")
  f.write(mon_texte)
  f.close()
```

### Écrire un fichier JSON

`write()` n'accepte que des strings : `f.write(mon_dict)` provoquerait une erreur. Pour enregistrer des listes et des
dictionnaires, on utilise `json.dump()`, qui fait l'inverse de `json.load()` : il prend les données Python et le fichier
ouvert en écriture.

- L'argument `indent=4` ajoute des retours à la ligne et des espaces pour que le fichier soit lisible par un humain.
- L'argument `ensure_ascii=False` garde les accents tels quels (sinon, `é` est écrit `\u00e9`).

```python
mon_dict = {
  "titre": "Les Misérables",
  "auteur": "Victor Hugo",
  "annee": 1862,
  "tomes": ["Fantine", "Cosette", "Marius", "L'Idylle rue Plumet et l'Épopée rue Saint-Denis", "Jean Valjean"]
}

chemin = os.path.join(racine, "sample_data", "mon_dict.json")

f = open(chemin, "w", encoding="utf-8")
json.dump(mon_dict, f, indent=4, ensure_ascii=False)
f.close()

# Vérifions :
f = open(chemin, encoding="utf-8")
print(f.read())
f.close()
```

### Écrire un fichier CSV

Enfin, pour écrire un tableau au format CSV, on utilise `csv.writer()`, qui prend en argument le fichier ouvert en
écriture. Sa méthode `writerow()` écrit une ligne, à partir d'une liste de cellules.

Notez l'argument `newline=""` dans `open()` : sans lui, sous Windows, une ligne vide apparaît entre chaque ligne du CSV.

```python
tableau = [
  ["titre", "auteur", "annee"],
  ["Les Misérables", "Victor Hugo", 1862],
  ["Germinal", "Émile Zola", 1885],
  ["Madame Bovary", "Gustave Flaubert", 1857]
]

chemin = os.path.join(racine, "sample_data", "nouveau_csv.csv")

f = open(chemin, "w", encoding="utf-8", newline="")
writer = csv.writer(f)
for ligne in tableau:
  writer.writerow(ligne)
f.close()

# Vérifions :
f = open(chemin, encoding="utf-8")
print(f.read())
f.close()
```

### Correction de l'exercice 3

```python
chemin = os.path.join(os.getcwd(), "sample_data", "README.md")
f = open(chemin, encoding="utf-8")
text = f.read()
f.close()

mon_dict = {}

for lettre in text:
  # D'abord, on vérifie qu'il s'agit bien d'une lettre (et pas d'un espace, d'un chiffre...) :
  if lettre.isalpha():
    # On met la lettre en minuscule, pour que "A" et "a" soient comptés ensemble :
    lettre_min = lettre.lower()

    if lettre_min in mon_dict:
      # La lettre a déjà été vue : on ajoute 1 à son compteur.
      mon_dict[lettre_min] = mon_dict[lettre_min] + 1
    else:
      # C'est la première fois qu'on voit cette lettre : on crée son compteur.
      mon_dict[lettre_min] = 1

print(mon_dict)
```
