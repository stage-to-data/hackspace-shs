## 1. Les conditions

Avec les fonctions et les variables, les conditions et les boucles forment la base de tout code en Python. Commençons
par les conditions.

Jusqu'ici, toutes les lignes de code que nous avons écrites ont été exécutées. Mais pour que notre code soit plus
dynamique, on peut vouloir exécuter certaines lignes seulement sous certaines conditions.

Prenez l'exercice proposé à la fin de la séance 1 : simuler une conversation avec l'ordinateur. Pour s'approcher d'une
vraie conversation, il faudrait que l'ordinateur réponde différentes choses en fonction de ce qu'on lui dit. Par
exemple, à la question « *Quel âge as-tu ?* », si l'utilisateur répond « *5 ans* », l'ordinateur pourrait répondre
« *Tu es jeune !* », et s'il répond « *100 ans* », « *Quelle expérience !* ».

Pour créer une condition, on utilise un **mot clé** : `if` (« si »). Pour commencer très simplement, vérifions si une
variable de type boolean vaut `True` (vrai) ou `False` (faux) :

```python
# D'abord, je crée une variable de type boolean :
my_bool = True

# Ici, j'écris ma condition. Je commence par le mot clé if, puis j'écris la condition :
# ici, la valeur de my_bool doit être égale à True.
if my_bool == True:
  # Voici ce qui se passe si la condition est remplie.
  # Comme pour une fonction, on a mis deux points et indenté la ligne suivante,
  # pour indiquer que ce qui suit appartient à la condition :
  print("C'est vrai")
```

Dans cet exemple, remplacez la valeur de `my_bool` par `False` : plus rien ne s'imprime, puisque la condition n'est plus
remplie ! Voilà notre première condition, et un code dynamique qui réagit aux données qu'on lui donne.

⚠ Remarquez qu'on utilise **deux** signes égal (`==`) pour *comparer* deux valeurs, alors qu'un seul signe égal (`=`)
sert à *donner* une valeur à une variable. C'est une erreur très fréquente !

### `else`

Continuons avec un autre mot clé : `else` (« sinon »). Dans l'exemple précédent, on a dit ce qui se passe quand la
condition est remplie ; quand elle ne l'est pas, il ne se passe rien. Ce serait bien de pouvoir faire autre chose dans
ce cas. On le fait avec un bloc `else`, à la suite du bloc `if` :

```python
my_bool = False

if my_bool == True:
  print("C'est vrai !")
else:
  print("C'est faux !")

# Notez que le bloc else se trouve juste à la suite du bloc if.
# On retrouve les deux points et le retour à la ligne indenté,
# mais on n'a pas besoin de donner une deuxième condition.
```

### Les opérateurs de comparaison

Nous avons vu des conditions qui vérifient si une variable est égale à une valeur donnée. Cela fonctionne avec
n'importe quel type de donnée, par exemple `if my_var == 50:` ou `if my_var == "bonjour":`. On peut aussi comparer
deux variables : `if my_var == une_autre_var:`.

Pour demander si quelque chose est **égal à** autre chose, on utilise l'**opérateur** `==`. Il en existe d'autres :

| Opérateur | Signification | Exemple vrai |
|---|---|---|
| `==` | égal à | `50 == 50` |
| `!=` | différent de | `50 != 25` |
| `<` | strictement plus petit que | `25 < 50` |
| `>` | strictement plus grand que | `50 > 25` |
| `<=` | plus petit que ou égal à | `50 <= 50` |
| `>=` | plus grand que ou égal à | `50 >= 50` |

Voyons cela en exemples :

```python
mon_nombre = 50

# On peut faire des conditions avec des nombres :
if mon_nombre == 50:
  print("C'est 50")
else:
  print("Ce n'est pas 50")

# Et utiliser les opérateurs < et > :
if mon_nombre > 25:
  print("plus grand que 25")
if mon_nombre < 100:
  print("plus petit que 100")

# On peut aussi utiliser >= et <= :
if mon_nombre >= 50:
  print("plus grand que ou égal à 50")

# Et savoir si quelque chose est différent d'autre chose avec != :
if mon_nombre != 25:
  print("n'est pas 25")
```

### `elif`

Voyons un troisième mot clé : `elif`. Pour l'instant, nous n'avons vu que des exemples où l'on vérifie **une**
condition, et où l'on fait éventuellement autre chose avec `else` si elle n'est pas remplie.

On peut enchaîner autant de conditions que l'on veut. Pour ajouter une deuxième condition, on procède comme avec `if`,
mais en écrivant `elif` (contraction de *else if*, « sinon, si »).

N'oubliez pas que l'ordinateur lit notre code dans l'ordre. Si la première condition (`if`) est remplie, il ignore
toutes les autres lignes du bloc. Sinon, il vérifie chaque `elif` dans l'ordre, jusqu'à en trouver un dont la condition
est remplie. Si aucune ne l'est, il exécute le `else` (s'il y en a un). L'**ordre** des conditions est donc important !

Voici un exemple :

```python
# On peut enchaîner les conditions avec elif (sinon, si...) :
mon_float = 15

# Ma condition initiale, avec if :
if mon_float >= 20:
  print("plus grand que ou égal à 20")

# Puis je peux ajouter autant de elif que je veux :
elif mon_float >= 16:
  print("entre 16 et 20")
elif mon_float >= 10:
  print("entre 10 et 16")

# Et si je veux, j'ajoute un else pour le cas où aucune condition n'est remplie :
else:
  print("plus petit que 10")
```

### Les conditions composées : `and` et `or`

Enfin, voyons comment composer des conditions plus complexes, avec deux derniers mots clés, à placer entre deux
conditions : `and` (« et ») et `or` (« ou »).

- Avec `and`, les deux conditions doivent **toutes les deux** être remplies pour que le code soit exécuté.
- Avec `or`, il suffit qu'**une seule** des deux conditions soit remplie.

Voici un exemple :

```python
mon_float = 15

# Ici, les deux conditions doivent être remplies :
if mon_float > 10 and mon_float < 20:
  print("entre 10 et 20")

# Ici, il suffit qu'une des deux conditions soit remplie :
if mon_float < 10 or mon_float > 20:
  print("en dehors de l'intervalle 10-20")
else:
  print("dans l'intervalle 10-20")

# Changez la valeur de mon_float pour voir ce qui change !
```

## 2. Les boucles

Bravo, nous y sommes presque ! Voici le dernier élément pour comprendre les bases de Python : les boucles.

Les boucles sont de nouveaux blocs que l'on peut mettre dans notre code pour exécuter des lignes de code plusieurs
fois. Cela peut sembler bizarre, mais nous allons voir que c'est très vite utile.

Prenons l'exemple ci-dessous :

```python
ma_liste = [10, 20, 30, 40, 50, 60]

print(ma_liste[0])
print(ma_liste[1])
print(ma_liste[2])
print(ma_liste[3])
print(ma_liste[4])
print(ma_liste[5])
```

Dans cet exemple, j'ai créé une liste de 6 éléments, puis j'ai voulu imprimer chacun d'eux. Pour cela, j'ai dû écrire 6
lignes, qui vont chercher chacune un élément de la liste. Cette approche pose plusieurs problèmes :

- C'est pénible ! Il y a beaucoup trop de lignes de code, et les copier-coller sont une source d'erreurs. Et ici, notre
  liste est petite : imaginez une liste de plusieurs centaines d'éléments !
- Et si le nombre d'éléments de la liste change ? S'il y en a moins, j'aurai une erreur, car j'essaierai d'aller
  chercher des éléments qui n'existent pas. S'il y en a plus, je ne les imprimerai pas.

### La boucle `for`

Heureusement, il existe une solution qui permet de faire la même chose de manière bien plus intelligente : la boucle
`for`. Voici un exemple :

```python
# Une boucle for permet de faire quelque chose avec chaque élément de la liste :
for item in ma_liste:
  print(item)
```

Décomposons la première ligne :
- la boucle commence par le mot clé `for` (« pour ») ;
- ensuite, on déclare une variable (on lui donne le nom que l'on veut, ici `item`) : c'est dans cette variable que sera
  placé, à chaque tour de boucle, l'élément de la liste ;
- puis le mot clé `in` (« dans ») : on est en train de dire « pour chaque élément dans... » ;
- et enfin, notre liste, contenue dans la variable `ma_liste`.

Comme pour une fonction ou une condition, on met deux points `:` et on indique le code qui appartient à la boucle avec
une indentation. Ce code est exécuté une fois pour chaque élément de la liste (à chaque « tour » de boucle, on parle
d'**itération**). Ici, on imprime simplement l'élément de la liste (contenu dans la variable `item`).

### Construire une liste avec une boucle : `len()` et `append()`

Deux outils vont très souvent avec les boucles :
- la fonction `len()` renvoie le nombre d'éléments d'une liste (ou le nombre de caractères d'un string) ;
- la **méthode** `append()` ajoute un élément à la fin d'une liste. Une méthode est une fonction qui appartient à une
  donnée : on l'appelle avec le nom de la variable, un point, puis le nom de la méthode : `ma_liste.append(element)`.

Très souvent, on part d'une liste vide `[]`, que l'on remplit au fur et à mesure dans une boucle :

```python
print(len(ma_liste))

# On veut une nouvelle liste contenant chaque élément de ma_liste multiplié par 2.
# On commence par une liste vide :
doubles = []

for item in ma_liste:
  # À chaque tour de boucle, on ajoute un élément à la fin de la liste :
  doubles.append(item * 2)

print(doubles)

# On peut aussi combiner boucles et conditions ! Ici, on ne garde que les éléments plus grands que 25 :
grands = []
for item in ma_liste:
  if item > 25:
    grands.append(item)

print(grands)
```

### `range()` et `enumerate()`

Nous savons maintenant faire des opérations pour chaque élément d'une liste. Mais il existe d'autres cas de figure. Et
si l'on voulait exécuter des lignes de code un certain nombre de fois, sans avoir de liste ?

Pour cela, on utilise, à la place d'une liste, une fonction qui s'appelle `range()`. `range(n)` produit la suite des
nombres de 0 à `n - 1` (donc `n` nombres au total), qui n'existe que le temps de la boucle. On peut donc faire ceci :

```python
nombre_diterations = 20

for i in range(nombre_diterations):
  print(i)
```

Ici, à la place d'une liste, on donne `range()`, avec comme argument le nombre de fois que l'on souhaite exécuter le
code. La variable `i` (encore une fois, on peut la nommer comme on veut) contient le numéro de l'itération en cours.
Remarquez que l'on compte à partir de 0, et que le dernier nombre est donc 19.

> 💡 `range()` peut aussi prendre un début et une fin : `range(1, 11)` va de 1 à 10 (la fin est toujours exclue).

Dernier cas de figure : on veut parcourir les éléments d'une liste, mais aussi connaître la position (l'**index**) de
l'élément en cours. Il existe pour cela une fonction qui s'appelle `enumerate()`. Voici un exemple :

```python
for index, nombre in enumerate(ma_liste):
  print(index, nombre)
```

Ici, à la place d'une liste ou de `range()`, on donne la fonction `enumerate()`, qui prend en argument la liste à
parcourir. À chaque tour de boucle, elle nous donne non seulement l'élément de la liste, mais aussi son index : c'est
pour cela qu'on déclare deux variables en même temps (`index` et `nombre`), séparées par une virgule.

### Parcourir un dictionnaire, et le mot clé `in`

On peut aussi parcourir un dictionnaire avec une boucle `for` : à chaque tour, on obtient une **clé**, que l'on peut
utiliser pour récupérer la valeur correspondante.

Le mot clé `in` sert aussi dans les conditions, pour savoir si un élément est dans une liste, ou si une clé existe dans
un dictionnaire :

```python
auteurs = {"Les Misérables": "Victor Hugo", "Germinal": "Émile Zola", "Madame Bovary": "Gustave Flaubert"}

for titre in auteurs:
  print(titre, "a été écrit par", auteurs[titre])

if "Germinal" in auteurs:
  print("Germinal est dans le dictionnaire")

if "Candide" not in auteurs:
  print("Candide n'est pas dans le dictionnaire")
```

### La boucle `while`

Enfin, voyons un autre type de boucle : `while` (« tant que »). Une boucle `while` combine en quelque sorte une boucle
et une condition : on lui donne une condition, et **tant que** cette condition est remplie, la boucle continue de
s'exécuter.

Voici un exemple :

```python
compteur = 0

while compteur < 10:
  print(compteur)
  compteur = compteur + 1
```

Nous avons déclaré notre boucle avec le mot clé `while`, suivi d'une condition : la variable `compteur` doit être
inférieure à 10. Tant que c'est le cas, la boucle s'exécute. C'est pour cela qu'à chaque tour de boucle, on ajoute 1 à
`compteur`.

⚠ **Attention** : si l'on ne faisait pas cela, `compteur` resterait toujours inférieur à 10, et la boucle ne
s'arrêterait jamais : c'est ce qu'on appelle une **boucle infinie**. Si cela vous arrive, cliquez sur le bouton ⏹ de la
cellule pour l'arrêter.

On utilise `while` quand on ne sait pas à l'avance combien de tours de boucle il faudra : par exemple, pour reposer une
question tant que l'utilisateur n'a pas donné une réponse valide.

### Le package `random`

Il existe énormément de packages, pour toutes sortes de tâches, et nous en verrons beaucoup dans les séances à venir.
Pour l'instant, commençons simplement, avec un package qui s'appelle `random`, et qui donne accès à tout un ensemble de
fonctions pour produire des données aléatoires. Il fait partie de Python : pas besoin de l'installer.

Commençons par importer ce package dans notre projet. Pour cela, on utilise le mot clé `import`, suivi du nom du
package :

```python
import random
```

Ensuite, pour utiliser les fonctions contenues dans ce package, on écrit le nom du package, un point, puis le nom de la
fonction. `random` contient par exemple une fonction `randint()`, qui renvoie un nombre entier aléatoire entre les deux
nombres qu'on lui donne (inclus). Voici comment l'utiliser :

```python
nombre_aleatoire = random.randint(100, 1000)

print(nombre_aleatoire)

# random.choice() choisit un élément au hasard dans une liste :
print(random.choice(["pierre", "feuille", "ciseaux"]))
```

Exécutez plusieurs fois cette cellule : le résultat change à chaque fois !
